Task: Write a function named urgent_tickets_sorted that accepts a list of customer-support ticket dictionaries. Each ticket has id, priority, created_at, and subject fields. Return only tickets whose priority is exactly "urgent", sorted newest to oldest by created_at. Dates use ISO 8601 format such as "2026-09-24T10:30:00".

Constraints:
- Use only built-in Python libraries.
- Return [] for an empty list.
- Do not modify the input list.
- Skip a ticket with a missing or invalid created_at value instead of raising an error.
- Return a list of the original ticket dictionaries.
- After the final code, provide a short explanation of how the edge cases are handled.

```python
from datetime import datetime


def urgent_tickets_sorted(tickets):
    """Return valid urgent tickets sorted newest to oldest by created_at."""
    dated_urgent_tickets = []

    for ticket in tickets:
        if not isinstance(ticket, dict) or ticket.get('priority') != 'urgent':
            continue

        try:
            created_at = datetime.fromisoformat(ticket.get('created_at'))
        except (TypeError, ValueError):
            continue

        dated_urgent_tickets.append((created_at, ticket))

    dated_urgent_tickets.sort(key=lambda item: item[0], reverse=True)
    return [ticket for _, ticket in dated_urgent_tickets]
```

```python
tickets = [
    {'id': 'T-101', 'priority': 'urgent', 'created_at': '2026-09-22T09:15:00', 'subject': 'Cannot sign in'},
    {'id': 'T-102', 'priority': 'normal', 'created_at': '2026-09-23T14:00:00', 'subject': 'Change profile name'},
    {'id': 'T-103', 'priority': 'urgent', 'created_at': '2026-09-24T10:30:00', 'subject': 'Dashboard is blank'},
    {'id': 'T-104', 'priority': 'urgent', 'created_at': 'not-a-date', 'subject': 'Payment did not go through'},
]

result = urgent_tickets_sorted(tickets)
print('Valid urgent tickets, newest first:')
for ticket in result:
    print(f"{ticket['id']} | {ticket['created_at']} | {ticket['subject']}")

print('\nEmpty list test:', urgent_tickets_sorted([]))
```
